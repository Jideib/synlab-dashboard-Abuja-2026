```python
def preclean_synlab_data():
    """Clean and standardize the SYNLAB survey data"""
    
    # Load the raw data with semicolon separator
    df = pd.read_csv('synlab_clean.csv', sep=';')
    
    # === RENAME COLUMNS TO STANDARD NAMES ===
    column_mapping = {
        '1. What is your age group?': 'age',
        '2. What is your gender?': 'gender',
        '3. What is your occupation?': 'occupation',
        '4. Kindly indicate the nearest location to you': 'location',
        '5. Do you have any form of health insurance or HMO coverage?': 'insurance',
        '6. Without looking at any list, which medical laboratory in Nigeria comes to your mind first?': 'top_of_mind_lab',
        '7. Which of the following medical laboratories in Nigeria are you aware of?/SYNLAB Nigeria': 'aware_synlab',
        '7. Which of the following medical laboratories in Nigeria are you aware of?/Clina Lancet': 'aware_clina_lancet',
        '7. Which of the following medical laboratories in Nigeria are you aware of?/Mecure Healthcare': 'aware_mecure',
        '7. Which of the following medical laboratories in Nigeria are you aware of?/Afriglobal Medicare': 'aware_afriglobal',
        '7. Which of the following medical laboratories in Nigeria are you aware of?/Clinix': 'aware_clinix',
        '7. Which of the following medical laboratories in Nigeria are you aware of?/AMCE (African Medical Centre of Excellence)': 'aware_amce',
        '7. Which of the following medical laboratories in Nigeria are you aware of?/Firmcare Diagnostics': 'aware_firmcare',
        '7. Which of the following medical laboratories in Nigeria are you aware of?/LAB360': 'aware_lab360',
        '7. Which of the following medical laboratories in Nigeria are you aware of?/APIN Medical Laboratory and Diagnostics': 'aware_apin',
        '7. Which of the following medical laboratories in Nigeria are you aware of?/Lifebridge Medical Diagnostics': 'aware_lifebridge',
        '7. Which of the following medical laboratories in Nigeria are you aware of?/Pseven Medical Diagnostics': 'aware_pseven',
        '7. Which of the following medical laboratories in Nigeria are you aware of?/e-Clinic & Diagnostics': 'aware_eclinic',
        '7. Which of the following medical laboratories in Nigeria are you aware of?/Echo Lab (EcoScan)': 'aware_echolab',
        '7. Which of the following medical laboratories in Nigeria are you aware of?/None of the above': 'aware_none',
        '9. Have you ever used the services of any of the following laboratories?/SYNLAB Nigeria': 'used_synlab',
        '9. Have you ever used the services of any of the following laboratories?/Clina Lancet': 'used_clina_lancet',
        '9. Have you ever used the services of any of the following laboratories?/Mecure Healthcare': 'used_mecure',
        '9. Have you ever used the services of any of the following laboratories?/Afriglobal Medicare': 'used_afriglobal',
        '9. Have you ever used the services of any of the following laboratories?/Clinix': 'used_clinix',
        '9. Have you ever used the services of any of the following laboratories?/Firmcare Diagnostics': 'used_firmcare',
        '9. Have you ever used the services of any of the following laboratories?/LAB360': 'used_lab360',
        '9. Have you ever used the services of any of the following laboratories?/APIN Medical Laboratory and Diagnostics': 'used_apin',
        '9. Have you ever used the services of any of the following laboratories?/Lifebridge Medical Diagnostics': 'used_lifebridge',
        '9. Have you ever used the services of any of the following laboratories?/Pseven Medical Diagnostics': 'used_pseven',
        '9. Have you ever used the services of any of the following laboratories?/e-Clinic & Diagnostics': 'used_eclinic',
        '9. Have you ever used the services of any of the following laboratories?/Echo Lab (EcoScan)': 'used_echolab',
        '9. Have you ever used the services of any of the following laboratories?/I have never used any medical laboratory': 'used_none',
        '11. When you need to use a medical laboratory, how do you typically decide which one to use?/I go to whichever laboratory my doctor recommends': 'decision_doctor',
        '11. When you need to use a medical laboratory, how do you typically decide which one to use?/I go to whichever laboratory my HMO covers': 'decision_hmo',
        '11. When you need to use a medical laboratory, how do you typically decide which one to use?/I choose based on proximity to my home or workplace': 'decision_proximity',
        '11. When you need to use a medical laboratory, how do you typically decide which one to use?/I choose based on previous experience': 'decision_experience',
        '11. When you need to use a medical laboratory, how do you typically decide which one to use?/I search online for options': 'decision_online',
        '11. When you need to use a medical laboratory, how do you typically decide which one to use?/I ask friends or family for a recommendation': 'decision_family',
        '11. When you need to use a medical laboratory, how do you typically decide which one to use?/I choose based on price': 'decision_price',
        '11. When you need to use a medical laboratory, how do you typically decide which one to use?/I choose based on the laboratory\'s reputation': 'decision_reputation',
        '12. On a scale of 1 to 5, how would you rate your overall impression of SYNLAB Nigeria?': 'overall_impression_score',
        '12b. On a scale of 0 to 10, how likely are you to recommend SYNLAB Nigeria to a friend, family member or colleague?': 'nps_score',
        '15a. Performance: Ease of access and location convenience': 'cx_access',
        '15b. Performance: Speed of result delivery': 'cx_result_speed',
        '15c. Performance: Quality and accuracy of results': 'cx_accuracy',
        '15d. Performance: Pricing and affordability': 'cx_value',
        '15e. Performance: Customer service and professionalism': 'cx_professionalism',
        '15f. Performance: Range of tests available': 'cx_range',
        '15g. Performance: Digital experience — online booking and result access': 'cx_digital',
        'Ease of finding and accessing the location': 'cx_access_alt',
        'Wait time at the facility': 'cx_wait_time',
        'Professionalism and courtesy of staff': 'cx_professionalism_alt',
        'Clarity of communication about the test and process': 'cx_communication',
        'Speed of receiving your results': 'cx_result_speed_alt',
        'Accuracy of your results': 'cx_accuracy_alt',
        'Online or digital experience if used': 'cx_digital_alt',
        'Overall value for the price paid': 'cx_value_alt',
        '18. If SYNLAB Nigeria offered a comprehensive annual health screening package, what price range would you consider fair value?': 'wtp_package',
        '17. When choosing a medical laboratory, how important is price in your decision?': 'price_importance',
        '19. How do you prefer to access your medical test results?': 'pref_result_access',
        '20. How would you prefer to book a laboratory appointment or test?': 'pref_booking',
        '_id': 'id',
    }
    
    # Rename only columns that exist in the dataframe
    existing_mapping = {k: v for k, v in column_mapping.items() if k in df.columns}
    df.rename(columns=existing_mapping, inplace=True)
    
    # === DATA TYPE CONVERSIONS ===
    # Convert binary columns (1.0/0.0) to integers
    binary_cols = [col for col in df.columns if col.startswith('aware_') or col.startswith('used_') or col.startswith('decision_')]
    for col in binary_cols:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0).astype(int)
    
    # Convert rating columns to numeric
    rating_cols = ['nps_score', 'overall_impression_score']
    for col in rating_cols:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors='coerce')
    
    # === CLEAN LOCATION NAMES ===
    if 'location' in df.columns:
        location_mapping = {
            'Wuse': 'Wuse',
            'Asokoro': 'Asokoro',
            'Gwarimpa': 'Gwarimpa',
            'Gwagwalada': 'Gwagwalada',
            'Kubwa': 'Kubwa',
            'Lugbe': 'Lugbe',
            'Jabi': 'Wuse',  # Consolidate Jabi into Wuse
            'Others': 'Other'
        }
        # Handle any variations
        df['location'] = df['location'].replace(location_mapping)
        # Standardize other locations
        df['location'] = df['location'].str.strip().str.title()
    
    # === CREATE NPS SEGMENT ===
    def get_nps_segment(score):
        if pd.isna(score):
            return None
        if score >= 9:
            return 'Promoter'
        elif score >= 7:
            return 'Passive'
        else:
            return 'Detractor'
    
    if 'nps_score' in df.columns:
        df['nps_segment'] = df['nps_score'].apply(get_nps_segment)
    
    # === CLEAN OCCUPATION ===
    if 'occupation' in df.columns:
        occ_mapping = {
            'Self-employed/business owner': 'Self-employed',
            'Corporate/salaried professional': 'Corporate Professional',
            'Civil servant': 'Civil Servant',
            'Healthcare professional': 'Healthcare Professional',
            'Student': 'Student',
            'Unemployed': 'Unemployed',
            'Other (please specify)': 'Other'
        }
        df['occupation'] = df['occupation'].replace(occ_mapping)
    
    # === CLEAN AGE GROUPS ===
    if 'age' in df.columns:
        age_order = ['Under 18', '18-24', '25-34', '35-44', '45-54', '55 and above']
        # Ensure consistency
        df['age'] = df['age'].str.strip()
    
    # === CLEAN INSURANCE ===
    if 'insurance' in df.columns:
        insurance_mapping = {
            'Yes, through my employer': 'Employer-provided',
            'Yes, I pay for it personally': 'Personal HMO',
            'No, I pay out of pocket': 'No insurance',
            'I am not sure': 'Not sure'
        }
        df['insurance'] = df['insurance'].replace(insurance_mapping)
    
    # === SAVE CLEANED DATA ===
    df.to_csv('synlab_clean_standardized.csv', index=False)
    print(f"✅ Data cleaned and saved! {len(df)} records processed.")
    print(f"📊 Columns available: {df.columns.tolist()}")
    
    return df

if __name__ == "__main__":
    preclean_synlab_data()
```
