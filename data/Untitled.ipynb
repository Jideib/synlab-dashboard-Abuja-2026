```python
"""
Simple deduplication script for SYNLAB survey data.
Removes duplicate rows based on content columns, keeping the first submission.
"""

import pandas as pd

# Load the data
df = pd.read_csv('synlab_clean.csv')

# Columns to exclude from duplicate checking (metadata only)
metadata_cols = ['id', '_uuid', '_submission_time', '_index', 'meta/rootUuid']

# Identify content columns (everything except metadata)
content_cols = [col for col in df.columns if col not in metadata_cols]

# Count rows before deduplication
print(f"Rows before: {len(df)}")

# Sort by submission time to keep earliest submission
if '_submission_time' in df.columns:
    df = df.sort_values('_submission_time')

# Remove duplicates based on content columns, keeping first occurrence
df_clean = df.drop_duplicates(subset=content_cols, keep='first')

# Count rows after deduplication
print(f"Rows after: {len(df_clean)}")
print(f"Rows removed: {len(df) - len(df_clean)}")

# Save cleaned data
df_clean.to_csv('synlab_clean_deduped.csv', index=False)
print("Saved to: synlab_clean_deduped.csv")
```
